# بررسی محیط اجرای پایتون (check_env)

این نوت‌بوک اطلاعات محیط اجرای پایتون را نمایش می‌دهد، از جمله: نسخه پایتون، مسیر اجرای پایتون، دایرکتوری کاری فعلی، نسخه کتابخانه‌های کلیدی و اطلاعات سیستم.

## ۱. نسخه پایتون

با استفاده از ماژول استاندارد `sys` نسخه پایتون نصب‌شده روی سیستم نمایش داده می‌شود.

```python
import sys
print("نسخه پایتون:", sys.version)
print("نسخه (خلاصه):", sys.version_info)
```

## ۲. دایرکتوری کاری فعلی

با ماژول `os` مسیر دایرکتوری که نوت‌بوک در آن اجرا می‌شود نمایش داده می‌شود.

```python
import os
print("دایرکتوری کاری فعلی:", os.getcwd())
```

## ۳. مسیر اجرای پایتون

مسیر فایل اجرایی مفسر پایتون که این نوت‌بوک با آن اجرا شده است.

```python
print("مسیر اجرای پایتون:", sys.executable)
```

## ۴. نسخه کتابخانه‌های کلیدی

نسخه کتابخانه‌های numpy، pandas، scikit-learn، torch و tensorflow (در صورت نصب بودن) بررسی می‌شود. اگر کتابخانه‌ای نصب نباشد، پیام مناسب نمایش داده می‌شود.

```python
import importlib

libraries = ["numpy", "pandas", "sklearn", "torch", "tensorflow"]
for lib in libraries:
    try:
        module = importlib.import_module(lib)
        version = getattr(module, "__version__", "نامشخص")
        print(f"{lib}: {version}")
    except ImportError:
        print(f"{lib}: نصب نشده است")
```

## ۵. اطلاعات سیستم

اطلاعات سیستم‌عامل و پلتفرم (نام سیستم‌عامل، نسخه، معماری پردازنده و نوع ماشین) با ماژول `platform` نمایش داده می‌شود.

```python
import platform

print("سیستم‌عامل:", platform.system())
print("نام سیستم‌عامل:", platform.os.name)
print("نسخه سیستم‌عامل:", platform.version())
print("انتشار:", platform.release())
print("معماری پردازنده:", platform.machine())
print("پردازنده:", platform.processor())
print("اطلاعات کامل پلتفرم:", platform.platform())
```
